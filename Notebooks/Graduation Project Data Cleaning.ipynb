# Data Cleaning & Processing

```python
# Converting Line DownTime Table From Wide to Long Format 
df_downtime_length = df_linedowntime.melt(
    id_vars=['Batch'], 
    var_name='Factor', 
    value_name='Duration'
).sort_values(by='Batch').reset_index(drop=True)

# Empty Cells Removal (Where There is No DownTime)
df_downtime_length = df_downtime_length.dropna(subset=['Duration'])

# Changing Column Name
df_downtime_length = df_downtime_length.rename(columns = {'Duration' : 'Downtime'})

# Now The Table Has 3 Columns : Batch, Factor, and Duration  
```

```python
# Dictionary Update with The New Table Format
dataframes = {
    "Line Productivity": df_lineproductivity,
    "Products": df_products,
    "Downtime Factors": df_downtimefactors,
    "Line Downtime": df_downtime_length # قمنا بتغييرها هنا لتشمل النسخة المرتبة
}
```

## Data Types Check

```python
# Loop to Display Data Types for Each Column in Each Table :
for name, df in dataframes.items():
    print(f"--- {name} ---")
    display(df.dtypes)
    print("\n" + "-"*30 + "\n")
```

## Data Types Casting

### Line Productivity Table 

#### 1 ) Start & End Time Columns

```python
# format='%H:%M:%S' To Tell Python How to Read The Time (ساعات:دقائق:ثواني)
df_lineproductivity['Start Time'] = pd.to_datetime(df_lineproductivity['Start Time'], format='%H:%M:%S').dt.time
df_lineproductivity['End Time'] = pd.to_datetime(df_lineproductivity['End Time'], format='%H:%M:%S').dt.time
```

```python
# Combining Both Date & Time Columns
df_lineproductivity['Start Datetime'] = pd.to_datetime(
    df_lineproductivity['Date'].astype(str) + ' ' + df_lineproductivity['Start Time'].astype(str)
)

df_lineproductivity['End Datetime'] = pd.to_datetime(
    df_lineproductivity['Date'].astype(str) + ' ' + df_lineproductivity['End Time'].astype(str)
)
# Original Columns Dropping
df_lineproductivity = df_lineproductivity.drop(columns=['Start Time', 'End Time'])
```

### Factors Table

#### 2 ) Operator Error Column

```python
# Converting into Booleans
df_downtimefactors['Operator Error'] = df_downtimefactors['Operator Error'].map({'Yes': True, 'No': False})
```

### Line DownTime Table 

#### 3 ) Factor Column

```python
# Converting into Integer
df_downtime_length['Factor'] = pd.to_numeric(df_downtime_length['Factor'], errors='coerce').astype(int)
```

#### 4 ) Duration Column 

```python
# Converting into Integer
df_downtime_length['Downtime'] = pd.to_numeric(df_downtime_length['Downtime'], errors='coerce').fillna(0).astype(int)
```

## Standardization

### Line Productivity Table

#### 1 : Date & Time Columns

```python
# 1. Create a mask: Identify rows where End time is 'earlier' than Start time
# This indicates the shift crossed midnight to the next day
midnight_cross_mask = df_lineproductivity['End Datetime'] < df_lineproductivity['Start Datetime']

# 2. Add one day to those specific End Datetime rows
df_lineproductivity.loc[midnight_cross_mask, 'End Datetime'] += pd.Timedelta(days=1)
```

### Products Table 

#### 2 : Product Size 

```python
# 1. Convert to string explicitly first using .astype(str)
# We add 'r' before the regex to ensure it's treated as a raw string
numeric_part = df_products['Size'].astype(str).str.extract(r'(\d+\.?\d*)').astype(float)

# 2. Add .str before .contains to avoid the next error
is_liters = df_products['Size'].astype(str).str.contains('l', case=False) & \
            ~df_products['Size'].astype(str).str.contains('ml', case=False)

# 3. Calculate the volume
import numpy as np
df_products['Size_ml'] = numeric_part[0] * np.where(is_liters, 1000, 1)
df_products['Size_ml'] = df_products['Size_ml'].astype(int)

# Dropping the original 'Size' column to keep the dataframe clean
df_products = df_products.drop(columns=['Size'])
```

## New Columns Creation

### Line Productivity Table

#### Batch Production Duration in Minutes

```python
# 3. calculate the duration now that the dates are accurate
df_lineproductivity['Manufacturing Time (Min)'] = ((df_lineproductivity['End Datetime'] - df_lineproductivity['Start Datetime']).dt.total_seconds() / 60).round().astype(int)
```

#### Batch Production Duration in Hours 

```python
df_lineproductivity['Manufacturing Time (Hrs)'] = (df_lineproductivity['Manufacturing Time (Min)'] / 60).round(2)
```

## Primary Keys Check 

```python
# Mapping the dataframes with their corresponding primary keys
pk_check_map = {
    "Line_Productivity": {'df': df_lineproductivity, 'pk': ['Batch']},
    "Products": {'df': df_products, 'pk': ['Product']},
    "Downtime_Factors": {'df': df_downtimefactors, 'pk': ['Factor']},
    "Downtime_Length": {'df': df_downtime_length, 'pk': ['Batch', 'Factor']}
}

# Iterate through the dictionary to validate keys
for name, data in pk_check_map.items():
    df = data['df']
    pk = data['pk']
    
    # Check for missing values in the primary key columns
    null_count = df[pk].isnull().sum().sum()
    
    # Check for duplicate entries
    dup_count = df.duplicated(subset=pk).sum()
    
    print(f"--- Checking table: {name} ---")
    print(f"Primary Key: {pk}")
    print(f"Null values: {null_count}")
    print(f"Duplicate values: {dup_count}")
    
    if null_count == 0 and dup_count == 0:
        print("Status: Valid and ready for merge.\n")
    else:
        print("Status: Warning! Key issues detected.\n")
```

# Data Modeling

```python
# 1. Join Line_Productivity (Fact) with Products (Dim)
# We join on 'Product' to get the product details for each batch
master_df = df_lineproductivity.merge(
    df_products, 
    on='Product', 
    how='left'
)

# 2. Join with Line_Downtime (Fact)
# We join on 'Batch' (Foreign Key in Line_Downtime pointing to Line_Productivity)
# This will align the downtime events with the specific production batch
master_df = master_df.merge(
    df_downtime_length, 
    on='Batch', 
    how='left'
)

# 3. Join with Downtime_Factors (Dim)
# We join on 'Factor' (Foreign Key in Line_Downtime pointing to Downtime_Factors)
# This replaces the cryptic factor code with the actual description
master_df = master_df.merge(
    df_downtimefactors, 
    on='Factor', 
    how='left'
)

# Final Validation
print("Master Dataset Shape:", master_df.shape)
print("Columns in Master Dataset:", master_df.columns.tolist())
display(master_df)
```

# Final Data Model 

```python
# 1. Fill numeric downtime values with 0
master_df['Downtime'] = master_df['Downtime'].fillna(0)

# 2. Fill text downtime values with "No Downtime"
master_df['Factor'] = master_df['Factor'].fillna('No Downtime')
master_df['Description'] = master_df['Description'].fillna('No Downtime')
master_df['Operator Error'] = master_df['Operator Error'].fillna('No Downtime')

# 3. Final verification to ensure no NaNs remain in these columns
print("Check for any remaining missing values:")
print(master_df[['Factor', 'Downtime', 'Description', 'Operator Error']].isnull().sum())

# Display a sample to see the clean data
display(master_df.sample(5))
```
